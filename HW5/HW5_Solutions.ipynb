## Problem Statement

Given dataset X (4×3):

$$X = \begin{bmatrix} 1 & 2 & 2 \\ 1 & 2 & 1 \\ 3 & 2 & 1 \\ 1 & 2 & 3 \end{bmatrix}$$

1. Compute the covariance matrix normalized by $(n-1)$, where $n$ is the number of features per sample.
2. How to select the projection matrix so that the projected covariance matrix is diagonal?
3. What are the principal components of X and their corresponding variances?

---
## Part 1: Covariance Matrix

Here $n = \text{number of features per sample} = 3$, so we normalize by $n-1 = 2$.

Since $n$ counts features **within** each sample, the covariance is computed within each sample: subtract each row's own mean, form the $3 \times 3$ outer product per row, then sum across all 4 samples and normalize by $n-1$:

$$C = \frac{1}{n-1}\, X_{rc}^T X_{rc} = \frac{1}{2}\, X_{rc}^T X_{rc}$$

where $X_{rc}$ is the **row-centered** data matrix.

### Step 1 — Row Means

For each sample $i$, compute $\mu_i = \dfrac{1}{n}\sum_{j=1}^{n} x_{ij}$ (mean across its 3 features):

$$\mu_1 = \frac{1+2+2}{3} = \frac{5}{3}, \qquad \mu_2 = \frac{1+2+1}{3} = \frac{4}{3}$$

$$\mu_3 = \frac{3+2+1}{3} = 2, \qquad \mu_4 = \frac{1+2+3}{3} = 2$$

### Step 2 — Row-Centered Data $X_{rc}$

Subtract each row's own mean:

$$X_{rc} = \begin{bmatrix}
1-\frac{5}{3} & 2-\frac{5}{3} & 2-\frac{5}{3} \\
1-\frac{4}{3} & 2-\frac{4}{3} & 1-\frac{4}{3} \\
3-2 & 2-2 & 1-2 \\
1-2 & 2-2 & 3-2
\end{bmatrix}
= \begin{bmatrix}
-\frac{2}{3} & \frac{1}{3} & \frac{1}{3} \\
-\frac{1}{3} & \frac{2}{3} & -\frac{1}{3} \\
1 & 0 & -1 \\
-1 & 0 & 1
\end{bmatrix}$$

**Verification:** every row sums to zero (as required by row-centering):

$$-\tfrac{2}{3}+\tfrac{1}{3}+\tfrac{1}{3}=0 \quad\checkmark \qquad -\tfrac{1}{3}+\tfrac{2}{3}-\tfrac{1}{3}=0 \quad\checkmark \qquad 1+0-1=0 \quad\checkmark \qquad -1+0+1=0 \quad\checkmark$$

### Step 3 — Scatter Matrix $X_{rc}^T X_{rc}$

Entry $[j,k] = \sum_{i=1}^{4} (X_{rc})_{ij}\,(X_{rc})_{ik}$. The matrix is symmetric; compute the 6 unique entries:

$$[1,1]: \left(-\tfrac{2}{3}\right)^2+\left(-\tfrac{1}{3}\right)^2+1^2+(-1)^2 = \tfrac{4}{9}+\tfrac{1}{9}+1+1 = \tfrac{5}{9}+\tfrac{18}{9} = \frac{23}{9}$$

$$[1,2]: \left(-\tfrac{2}{3}\right)\left(\tfrac{1}{3}\right)+\left(-\tfrac{1}{3}\right)\left(\tfrac{2}{3}\right)+(1)(0)+(-1)(0) = -\tfrac{2}{9}-\tfrac{2}{9} = -\frac{4}{9}$$

$$[1,3]: \left(-\tfrac{2}{3}\right)\left(\tfrac{1}{3}\right)+\left(-\tfrac{1}{3}\right)\left(-\tfrac{1}{3}\right)+(1)(-1)+(-1)(1) = -\tfrac{2}{9}+\tfrac{1}{9}-1-1 = -\tfrac{1}{9}-2 = -\frac{19}{9}$$

$$[2,2]: \left(\tfrac{1}{3}\right)^2+\left(\tfrac{2}{3}\right)^2+0^2+0^2 = \tfrac{1}{9}+\tfrac{4}{9} = \frac{5}{9}$$

$$[2,3]: \left(\tfrac{1}{3}\right)\left(\tfrac{1}{3}\right)+\left(\tfrac{2}{3}\right)\left(-\tfrac{1}{3}\right)+(0)(-1)+(0)(1) = \tfrac{1}{9}-\tfrac{2}{9} = -\frac{1}{9}$$

$$[3,3]: \left(\tfrac{1}{3}\right)^2+\left(-\tfrac{1}{3}\right)^2+(-1)^2+1^2 = \tfrac{1}{9}+\tfrac{1}{9}+1+1 = \tfrac{2}{9}+\tfrac{18}{9} = \frac{20}{9}$$

$$X_{rc}^T X_{rc} = \begin{bmatrix} \frac{23}{9} & -\frac{4}{9} & -\frac{19}{9} \\[4pt] -\frac{4}{9} & \frac{5}{9} & -\frac{1}{9} \\[4pt] -\frac{19}{9} & -\frac{1}{9} & \frac{20}{9} \end{bmatrix}$$

### Step 4 — Covariance Matrix $C = \dfrac{1}{2}\, X_{rc}^T X_{rc}$

$$\boxed{C = \begin{bmatrix} \dfrac{23}{18} & -\dfrac{2}{9} & -\dfrac{19}{18} \\[8pt] -\dfrac{2}{9} & \dfrac{5}{18} & -\dfrac{1}{18} \\[8pt] -\dfrac{19}{18} & -\dfrac{1}{18} & \dfrac{10}{9} \end{bmatrix} \approx \begin{bmatrix} 1.278 & -0.222 & -1.056 \\ -0.222 & 0.278 & -0.056 \\ -1.056 & -0.056 & 1.111 \end{bmatrix}}$$

**Key difference from column-centering:** feature 2 no longer has zero variance ($C_{22} = \frac{5}{18} \neq 0$), because each sample's value of feature 2 can still deviate from *that sample's own mean*, even though feature 2 is constant across samples.

**Trace check:** $\mathrm{tr}(C) = \dfrac{23}{18}+\dfrac{5}{18}+\dfrac{10}{9} = \dfrac{23+5+20}{18} = \dfrac{48}{18} = \dfrac{8}{3}$

---
## Part 2: Selecting the Projection Matrix

**The projection matrix $W$ must have the eigenvectors of $C$ as its rows.**

Since $C$ is real and symmetric, it has an orthogonal eigendecomposition $C = V\Lambda V^T$, where $V$ has eigenvectors as columns and $\Lambda$ is diagonal with eigenvalues. Setting $W = V^T$ (rows of $W$ = eigenvectors of $C$):

$$C_{\text{proj}} = W C W^T = V^T \cdot (V\Lambda V^T) \cdot V = \Lambda \quad \text{(diagonal)}$$

**Note on dimensions:** $W$ remains $3\times 3$ for the full decomposition. After projecting, any PC whose eigenvalue is zero contributes a zero-variance dimension that is discarded from the *projected data*, not from $W$ itself.

## Part 3: Principal Components and Their Variances

Find eigenvalues and eigenvectors of $C$ by solving $\det(C - \lambda I) = 0$.

### Eigenvalue 3 — Null Vector by Structure

Because every row of $X_{rc}$ sums to zero (row-centering constraint), we have $X_{rc}\,[1,1,1]^T = \mathbf{0}$, which implies:

$$C\,[1,1,1]^T = \tfrac{1}{2}\,X_{rc}^T \underbrace{X_{rc}\,[1,1,1]^T}_{=\,\mathbf{0}} = \mathbf{0}$$

**Direct verification:**

$$\text{Row 1: } \tfrac{23}{18} - \tfrac{2}{9} - \tfrac{19}{18} = \tfrac{23-4-19}{18} = 0\;\checkmark$$

$$\text{Row 2: } -\tfrac{2}{9}+\tfrac{5}{18}-\tfrac{1}{18} = \tfrac{-4+5-1}{18} = 0\;\checkmark$$

$$\text{Row 3: } -\tfrac{19}{18}-\tfrac{1}{18}+\tfrac{10}{9} = \tfrac{-19-1+20}{18} = 0\;\checkmark$$

Therefore:
$$\lambda_3 = 0, \qquad \mathbf{w}_3 = \frac{1}{\sqrt{3}}\begin{bmatrix}1\\1\\1\end{bmatrix}$$

### Eigenvalues 1 and 2 — Characteristic Polynomial

Since $\lambda_3 = 0$, the characteristic polynomial factors as $\lambda(\lambda^2 - \mathrm{tr}(C)\,\lambda + \sigma) = 0$, where $\sigma$ is the sum of the three $2\times 2$ principal minors of $C$.

**Sum of $2\times 2$ principal minors** (all three are equal by the matrix's structure):

$$M_{11} = \det\begin{bmatrix}\frac{5}{18}&-\frac{1}{18}\\-\frac{1}{18}&\frac{10}{9}\end{bmatrix} = \frac{5}{18}\cdot\frac{10}{9} - \left(\frac{1}{18}\right)^2 = \frac{50}{162}-\frac{1}{324} = \frac{100-1}{324} = \frac{99}{324} = \frac{11}{36}$$

$$M_{22} = \det\begin{bmatrix}\frac{23}{18}&-\frac{19}{18}\\-\frac{19}{18}&\frac{10}{9}\end{bmatrix} = \frac{23}{18}\cdot\frac{10}{9} - \left(\frac{19}{18}\right)^2 = \frac{460-361}{324} = \frac{99}{324} = \frac{11}{36}$$

$$M_{33} = \det\begin{bmatrix}\frac{23}{18}&-\frac{2}{9}\\-\frac{2}{9}&\frac{5}{18}\end{bmatrix} = \frac{23}{18}\cdot\frac{5}{18} - \left(\frac{2}{9}\right)^2 = \frac{115-16}{324} = \frac{99}{324} = \frac{11}{36}$$

$$\sigma = M_{11}+M_{22}+M_{33} = 3\cdot\frac{11}{36} = \frac{11}{12}$$

**Reduced quadratic** (dividing out $\lambda = 0$, using $\mathrm{tr}(C) = 8/3$):

$$\lambda^2 - \frac{8}{3}\lambda + \frac{11}{12} = 0 \quad\xrightarrow{\times\,12}\quad 12\lambda^2 - 32\lambda + 11 = 0$$

**Discriminant:**

$$\Delta = 32^2 - 4(12)(11) = 1024 - 528 = 496 = 16\cdot 31 \implies \sqrt{\Delta} = 4\sqrt{31}$$

$$\lambda = \frac{32 \pm 4\sqrt{31}}{24} = \frac{8 \pm \sqrt{31}}{6}$$

$$\lambda_1 = \frac{8+\sqrt{31}}{6} \approx \frac{8+5.568}{6} \approx \mathbf{2.261}, \qquad \lambda_2 = \frac{8-\sqrt{31}}{6} \approx \frac{8-5.568}{6} \approx \mathbf{0.405}$$

**Verification:**
$$\lambda_1+\lambda_2 = \frac{16}{6} = \frac{8}{3} = \mathrm{tr}(C)\;\checkmark \qquad \lambda_1\lambda_2 = \frac{64-31}{36} = \frac{33}{36} = \frac{11}{12} = \sigma\;\checkmark$$

### Eigenvectors for $\lambda_1$ and $\lambda_2$

Since $\mathbf{w}_3 = [1,1,1]^T/\sqrt{3}$ spans the null space, every other eigenvector must satisfy $w_1+w_2+w_3 = 0$, i.e., $w_1 = -w_2 - w_3$. Substituting into Row 1 of $(C-\lambda I)\mathbf{w}=0$:

$$\left(\frac{23}{18}-\lambda\right)(-w_2-w_3) - \frac{2}{9}w_2 - \frac{19}{18}w_3 = 0$$

After simplification this reduces to:

$$\left(\sqrt{31}-1\right)w_2 + \left(\sqrt{31}-6\right)w_3 = 0 \quad\text{(for }\lambda_1\text{)} \qquad \left(1+\sqrt{31}\right)w_2 + \left(6+\sqrt{31}\right)w_3 = 0 \quad\text{(for }\lambda_2\text{)}$$

**For $\lambda_1 = (8+\sqrt{31})/6 \approx 2.261$:**

$$\|\mathbf{v}_1\| = \sqrt{6.568^2+0.568^2+6^2} = \sqrt{43.14+0.32+36} = \sqrt{79.46} \approx 8.914$$

$$\boxed{\mathbf{w}_1 \approx [0.737,\ -0.064,\ -0.673]}$$

**For $\lambda_2 = (8-\sqrt{31})/6 \approx 0.405$:**

**Eigenvalue verification** $C\mathbf{w}_1 = \lambda_1\mathbf{w}_1$:
